```python
from transformers import pipeline
```

```python
llm = pipeline("text2text-generation",
        model="google/flan-t5-base",
        max_new_tokens=100)
```

```python
topic = "Tokyo"
prompt = f"""
    Give me 1 interesting sentence about {topic}
    then give me 1 fun fact about {topic}"""

    #Each fact should be one sentence only, numbered 1 to 3."
    #"""

response = llm(prompt)#[0]['generated_text']
```

```python
from langchain_community.llms import HuggingFacePipeline
```

```python
def build_fact_generator():
    model_name = "google/flan-t5-small"  # Lightweight instruct model
    generator = pipeline("text2text-generation", model=model_name, max_new_tokens=100)
    return HuggingFacePipeline(pipeline=generator)

llm = build_fact_generator()
prompt = f"Give me 3 interesting facts about {topic}. Each fact should be one sentence only, numbered 1 to 3."
response = llm.invoke(prompt)
```
