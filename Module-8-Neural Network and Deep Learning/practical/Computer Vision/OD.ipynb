```python
# Splitting Image
red, green, blue = cv2.split(new_image)
print("Red",red.shape)
print("Blue",blue.shape)
print("green",green.shape)

#merge image
new_image1 = cv2.merge([red, green, blue])
plt.imshow(blue)
```

```python
(h, w) = new_image.shape[:2]
print(h,w)

center = (w//2, h//2)
print(center)

angle = 180

m = cv2.getRotationMatrix2D(center, angle, 1.0)
Rotated90 = cv2.warpAffine(new_image, m, (h, w))
plt.imshow(Rotated90)
```
