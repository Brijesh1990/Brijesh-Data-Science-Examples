```python
def cal(x,y,z):
  if z == "+":
    p = x + y
  elif z == "-":
    p = x - y
  elif z == "*":
    p = x * y
  elif z == "/":
    p = x / y
  else:
    p = "Error"
  return p
```

```python
class Bank:
  def __init__(self,Name,Balance):
    self.Name = Name
    self.Balance = Balance
  def show_details(self):
    print(f"Name is {self.Name} and Balance is {self.Balance}")

class Tran(Bank):
  def deposit(self,amount):
    print(f"Your Deposite amt is {amount} and your current Balance is {self.Balance + amount}")
  def withdraw(self,amount):
    print(f"Your Deposite amt is {amount} and your current Balance is {self.Balance - amount}")
```
